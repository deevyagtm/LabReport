```python
import numpy as np
from collections import Counter

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]

mean_X = np.mean(X)
median_X = np.median(X)
mode_X = Counter(X).most_common(1)[0][0]

print("Mean of X =", mean_X)
print("Median of X =", median_X)
print("Mode of X =", mode_X)
```

```python
import numpy as np

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]

variance_X = np.var(X)
std_dev_X = np.std(X)

print("Variance of X =", variance_X)
print("Standard Deviation of X =", std_dev_X)
```

```python
import numpy as np

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]
Y = [35, 50, 70, 40, 85, 58, 75, 44, 90, 65]

covariance = np.cov(X, Y, bias=True)[0][1]

print("Covariance between X and Y =", covariance)
```

```python
import numpy as np

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]
Y = [35, 50, 70, 40, 85, 58, 75, 44, 90, 65]

correlation = np.corrcoef(X, Y)[0][1]

print("Correlation Coefficient =", correlation)
```

```python
import matplotlib.pyplot as plt

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]

plt.hist(X, bins=10)
plt.xlabel("Values of X")
plt.ylabel("Frequency")
plt.title("Histogram of X")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]

plt.boxplot(X)
plt.title("Box Plot of X")
plt.ylabel("Values")
plt.show()

Q1 = np.percentile(X, 25)
Q2 = np.percentile(X, 50)
Q3 = np.percentile(X, 75)

print("Q1 (25%) =", Q1)
print("Median (Q2) =", Q2)
print("Q3 (75%) =", Q3)
print("Outliers: Values outside whiskers (if any).")
```

```python
import matplotlib.pyplot as plt

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]
Y = [35, 50, 70, 40, 85, 58, 75, 44, 90, 65]

plt.scatter(X, Y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Scatter Plot of X vs Y")
plt.show()
```

```python
import pandas as pd

X = [23, 45, 67, 34, 89, 56, 78, 41, 92, 60]
Y = [35, 50, 70, 40, 85, 58, 75, 44, 90, 65]

data = pd.DataFrame({"X": X, "Y": Y})
corr_matrix = data.corr()

print("Correlation Matrix:")
print(corr_matrix)

value = corr_matrix.loc["X", "Y"]

if value > 0.7:
    print("Strong positive correlation")
elif value < -0.7:
    print("Strong negative correlation")
else:
    print("Weak or moderate correlation")
```
